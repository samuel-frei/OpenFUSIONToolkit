```python
import h5py
import numpy as np

# Open the file in read/write mode
filename = 'paper_eq_0115.h5'

with h5py.File(filename, 'r+') as f:
    # Access the dataset
    mesh_r = f['mesh/R']
    
    # Read the data
    data = mesh_r[:]
    
    # Remove the last column (keep only first 2 columns)
    data_modified = data[:, :2]
    
    # Delete the old dataset
    del f['mesh/R']
    
    # Create new dataset with modified data
    f.create_dataset('mesh/R', data=data_modified)

print(f"Modified 'mesh/R' from shape {data.shape} to {data_modified.shape}")
print("File saved successfully")
```
